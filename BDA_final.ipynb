```python
import pandas as pd
import os as os
import warnings
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import glob
import os
import numpy as np
warnings.filterwarnings("ignore")

# Here we are merging our major data
csv_files = ["Mumbai.csv", "Bangalore.csv", "Hyderabad.csv", "Kolkata.csv", "Delhi.csv", "Chennai.csv"]


dfs = [] # dfs here i have created a dataframe list for adding the daatasets

for file in csv_files:

    if os.path.exists(file):
        city_name = os.path.splitext(os.path.basename(file))[0]
        df = pd.read_csv(file) # reads file each city data becomes a dataframe
        df["city_name"] = city_name
        dfs.append(df) # using list cause pd.concat() can merge by doing it once



if dfs:
    merged_df = pd.concat(dfs, ignore_index=True) # merging

    merged_df.to_csv("merged_cities_data.csv", index=False) # saving it
    print("✅ Merged CSV saved as 'merged_cities_data.csv'")


    display(merged_df.head())
```

```python
# Calculate missing 'rate_persqft' values where possible
# Ensuring prise and area to be numeric
House_price_final_df['Price'] = pd.to_numeric(House_price_final_df['Price'], errors='coerce')
House_price_final_df['Area'] = pd.to_numeric(House_price_final_df['Area'], errors='coerce')

# Calculate rate_persqft only for rows where it is currently null and Area is not zero
null_rate_persqft_mask = House_price_final_df['rate_persqft'].isnull()
non_zero_area_mask = House_price_final_df['Area'].notnull() & (House_price_final_df['Area'] != 0)

House_price_final_df.loc[null_rate_persqft_mask & non_zero_area_mask, 'rate_persqft'] = (
    House_price_final_df['Price'] / House_price_final_df['Area']
)

# Convert 'rate_persqft' to integer, coercing errors to NaN and then filling NaNs with 0
House_price_final_df['rate_persqft'] = pd.to_numeric(House_price_final_df['rate_persqft'], errors='coerce').fillna(0).astype(int)

print("✅ Calculated and filled missing 'rate_persqft' values.")
display(House_price_final_df[columns_to_check].count()) # Re-check non-null count
```

Finding most common facelities

```python
#Top 5 most common facilities:

facility_count = {}

# Loop through each column
for column in house_data_cleaned.columns:
    # Check if the column represents a facility
    if set(house_data_cleaned[column].unique()) == {0.0, 1.0}:
        # If so, count the number of properties that have this facility
        count = house_data_cleaned[column].sum()
        # Add the count to the dictionary
        facility_count[column] = count

# Sort the dictionary by the count of each facility in descending order
sorted_facilities = sorted(facility_count.items(), key=lambda x: x[1], reverse=True)

# Print the top 5 most common facilities
print("Top 5 most common facilities:")
for facility, count in sorted_facilities[:5]:
    print(f"{facility}: {count} properties")
```

```python
# Create a bar plot of the top 5 most common facilities
top_facilities = dict(sorted_facilities[:5])
fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(x=list(top_facilities.keys()), y=list(top_facilities.values()), ax=ax, palette='viridis')
plt.xlabel("Facility")
plt.ylabel("Number of Properties")
plt.title("Most Common Facilities in Properties")
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels overlapping
plt.show()
```

Top Locations

 Top 3 Localities In Every City where this Basic Amenities is availiable at affordable Price.

```python
import matplotlib.ticker as mtick
import ipywidgets as widgets
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


def get_top_affordable_locations(city):


    city_data = house_data_cleaned[house_data_cleaned["city_name"] == city]

    filtered_data = city_data[(city_data["LiftAvailable"] == 1) &
                              (city_data["PowerBackup"] == 1) &
                              (city_data["Gymnasium"] == 1) &
                              (city_data["Children'splayarea"] == 1) &
                              (city_data["CarParking"] == 1)]

    # Group the data by location and calculate the average price
    location_stats = filtered_data.groupby("Location")["Price"].mean()

    # Get the top 3 locations based on the minimum average price
    top_locations = location_stats.nsmallest(3)
    print(f"Top 3 most affordable locations in {city} with basic amenities:\n{top_locations}")

    sns.set_style("whitegrid")
    ax = sns.barplot(x=top_locations.index, y=top_locations, palette="viridis")
    ax.set(xlabel="Location", ylabel="Average Price (Lakhs)", title=f"Top 3 Locations in {city} with Basic Amenities ")

    # Format y-axis labels in millions
    ax.yaxis.set_major_formatter(mtick.FuncFormatter(lambda x, pos: f"{x:.1f}L"))

    plt.show()

get_top_affordable_locations('Mumbai')
```


Why this insight is useful?

If you are someone who is plannning to move in a new city and looking the place where you can get all the basic amenities at low price,Then this data can be useful.

```python
# Dynamic Graphical Representation
w = widgets.interactive(get_top_affordable_locations,
city = widgets.Dropdown(options=['Mumbai', 'Delhi', 'Chennai', 'Hyderabad', 'Kolkata', 'Bangalore'],value='Mumbai',
                                    description='city:',disabled=False,
                         continuous_update=True,readout=True))

display(w)
```

How do amenities affect prices ?




```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
# pca
#StandardScaler to normalize them
#PCA to reduce them to 2 dimensions

pca_pipeline = Pipeline(steps = [('standard_scaler',StandardScaler()),
                                 ('pca',PCA(n_components=2))
                                 ]
                       )



amenities_cols = ['MaintenanceStaff', 'Gymnasium', 'SwimmingPool', 'LandscapedGardens',
       'JoggingTrack', 'RainWaterHarvesting', 'IndoorGames', 'ShoppingMall',
       'Intercom', 'SportsFacility', 'ATM', 'ClubHouse', 'School',
       '24X7Security', 'PowerBackup', 'CarParking', 'StaffQuarter',
       'Cafeteria', 'MultipurposeRoom', 'Hospital', 'WashingMachine',
       'Gasconnection', 'AC', "Children'splayarea", 'LiftAvailable',
       'BED', 'VaastuCompliant', 'Microwave', 'GolfCourse', 'TV',
       'DiningTable', 'Sofa', 'Wardrobe']


amenities_pc = pd.DataFrame(pca_pipeline.fit_transform(house_data_cleaned[amenities_cols]),
                            columns = [f'dim{i}' for i in range(1,3)])

price = house_data_cleaned.Price.values
amenities_pc['price'] = price

#city
amenities_pc['city'] = house_data_cleaned['city_name'].values


color_map_1 = dict(zip(amenities_pc.city.unique(),['r','g','b','orange','black','brown']))
amenities_pc['color'] = amenities_pc['city'].map(color_map_1)

# Filter data for price up to 2500 Lakhs
filtered_amenities_pc = amenities_pc[amenities_pc['price'] <= 2500].copy()


plt.scatter(filtered_amenities_pc['dim1'],filtered_amenities_pc['price'],color=filtered_amenities_pc['color'],s=5)

plt.xlabel('Principal Component of Amenities')
plt.ylabel('House Price in lakh(i.e 100 Thousand) Rupees)')
plt.title("Distribution of House prices",font="Serif", size=15,color='g')
plt.legend(handles=[plt.Line2D([], [], marker='o', color='w', markerfacecolor=color, label=city, markersize=10) for city, color in color_map_1.items()], title='City') # Added legend
plt.show()
```
